#1. Análisis de facturación

En este primer análisis se calcula el ingreso total generado por cada tienda.

Para ello se suman los valores de la columna **Precio** de cada conjunto de datos. Esto permite estimar cuánto dinero generó cada tienda a partir de sus ventas.

Este análisis es importante para comparar el desempeño de las tiendas y entender cuál genera mayores ingresos.

```python
facturacion_tienda1 = tienda["Precio"].sum()
facturacion_tienda2 = tienda2["Precio"].sum()
facturacion_tienda3 = tienda3["Precio"].sum()
facturacion_tienda4 = tienda4["Precio"].sum()

print("Facturación Tienda 1:", facturacion_tienda1)
print("Facturación Tienda 2:", facturacion_tienda2)
print("Facturación Tienda 3:", facturacion_tienda3)
print("Facturación Tienda 4:", facturacion_tienda4)
```

```python
import pandas as pd

facturacion = pd.DataFrame({
    "Tienda": ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"],
    "Facturación": [
        facturacion_tienda1,
        facturacion_tienda2,
        facturacion_tienda3,
        facturacion_tienda4
    ]
})

facturacion
```

# Interpretación

A partir del análisis de facturación se puede observar el ingreso total generado por cada tienda.

Esta información permite identificar qué tienda presenta el mayor volumen de ventas y comparar el desempeño entre ellas.

#4. Productos más y menos vendidos

En este análisis se identifican los productos más vendidos y los menos vendidos en cada tienda.

Para ello se cuenta la cantidad de veces que aparece cada producto en el conjunto de datos. Esto permite conocer cuáles son los artículos con mayor demanda y cuáles presentan menor volumen de ventas.

```python
ventas_productos_tienda1 = tienda["Producto"].value_counts()

producto_mas_vendido_tienda1 = ventas_productos_tienda1.idxmax()
producto_menos_vendido_tienda1 = ventas_productos_tienda1.idxmin()

print("Producto más vendido Tienda 1:", producto_mas_vendido_tienda1)
print("Producto menos vendido Tienda 1:", producto_menos_vendido_tienda1)
```

```python
ventas_productos_tienda2 = tienda2["Producto"].value_counts()

producto_mas_vendido_tienda2 = ventas_productos_tienda2.idxmax()
producto_menos_vendido_tienda2 = ventas_productos_tienda2.idxmin()

print("Producto más vendido Tienda 2:", producto_mas_vendido_tienda2)
print("Producto menos vendido Tienda 2:", producto_menos_vendido_tienda2)
```

```python
ventas_productos_tienda3 = tienda3["Producto"].value_counts()

producto_mas_vendido_tienda3 = ventas_productos_tienda3.idxmax()
producto_menos_vendido_tienda3 = ventas_productos_tienda3.idxmin()

print("Producto más vendido Tienda 3:", producto_mas_vendido_tienda3)
print("Producto menos vendido Tienda 3:", producto_menos_vendido_tienda3)
```

```python
ventas_productos_tienda4 = tienda4["Producto"].value_counts()

producto_mas_vendido_tienda4 = ventas_productos_tienda4.idxmax()
producto_menos_vendido_tienda4 = ventas_productos_tienda4.idxmin()

print("Producto más vendido Tienda 4:", producto_mas_vendido_tienda4)
print("Producto menos vendido Tienda 4:", producto_menos_vendido_tienda4)
```

## Interpretación

El análisis de productos más y menos vendidos permite identificar qué artículos tienen mayor aceptación entre los clientes y cuáles presentan menor demanda.

Esta información puede ser útil para ajustar estrategias de inventario, promoción o rotación de productos en cada tienda.
